```python
import numpy as np
```

```python
from matplotlib  import pyplot as plt
```

```python
x= np.random.randint(10,50,10)
y= np.random.randint(10,50,10)
z= np.random.randint(10,50,10)
```

```python
corr=np.corrcoef([x,y,z])
```

```python
corr
```

```python
corr.T
```

```python
plt.imshow(corr,cmap="coolwarm")
```

```python
plt.imshow(corr,cmap="coolwarm")
plt.xticks([0,1,2],labels=["Age","height","Weight"])
plt.yticks([0,1,2],labels=["Age","height","Weight"])
```

```python
plt.imshow(corr,cmap="coolwarm")
plt.xticks([0,1,2],labels=["Age","height","Weight"])
plt.yticks([0,1,2],labels=["Age","height","Weight"])
plt.text(0,0,corr[0,0],va="center",ha="center",c="white")
plt.text(1,0,round(corr[0,1],2),va="center",ha="center",c="white")
```

```python
plt.imshow(corr,cmap="coolwarm")
plt.xticks([0,1,2],labels=["Age","height","Weight"])
plt.yticks([0,1,2],labels=["Age","height","Weight"])
for x in range(3):
    for y in range(3):
        # print(x,y)
        plt.text(x,y,round(corr[y,x],2),va="center",ha="center",c="white")
```

```python
plt.imshow(corr,cmap="coolwarm")
values =range(3)
labels= ["Age","height","Weight"]
plt.xticks(values,labels=labels)
plt.yticks(values,labels=labels)
for x in values:
    for y in values:
        # print(x,y)
        plt.text(x,y,round(corr[y,x],2),va="center",ha="center",c="white")
```
