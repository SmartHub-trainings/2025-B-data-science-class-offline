```python
import  numpy as np
```

```python
np.random.seed(4)
scores= np.random.randint(0,101,(5,3))
```

```python
scores
```

```python
scores.mean()
```

```python
scores.mean(axis=1)
```

```python
avg_courses=scores.mean(axis=0)
```

```python
courses= ["English","Maths","Physics"]
```

```python
avg_courses
```

```python
from matplotlib import pyplot as plt
```

```python
plt.bar(courses,avg_courses)
```

```python
plt.pie(avg_courses);
```

```python
plt.pie(avg_courses,labels=courses);
```

```python
plt.pie(avg_courses,labels=courses,autopct="%.1f%%",explode=[0.1,0,0]);
```
