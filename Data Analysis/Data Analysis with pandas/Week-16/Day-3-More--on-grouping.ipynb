```python
import pandas as pd
```

```python
df=pd.read_csv("../datasets/SS_meridian.csv")
```

```python
p_class_grouping=df.groupby("Pclass")
```

```python
p_class_grouping["Fare"].mean()
```

```python
p_class_grouping["Fare"].sum()
```

```python
p_class_grouping["Fare"].count()
```

```python
p_class_grouping["Fare"].agg(["sum","mean","count"])
```

```python
p_class_grouping.agg(
    total_fare=("Fare","mean"),
    mean_fare=("Fare","sum"),
    no_of_individuals=("Fare","count")
)
```

```python
p_class_grouping.agg(
    avg_fare=("Fare","mean"),
    no_of_tickets=("Ticket","count"),
    avg_age=("Age","mean")
)
```
