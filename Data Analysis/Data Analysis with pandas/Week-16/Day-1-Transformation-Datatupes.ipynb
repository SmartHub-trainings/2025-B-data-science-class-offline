```python
def optimize_data_memory(df):
    columns = df.columns
    for column in columns:
        if str(df[column].dtype)[:2]=="in":
            max_int=df[column].max()
            min_int = df[column].min()
            if max_int<127 and min_int>-127:
                df[column] = df[column].astype("int8")
        elif str(df[column].dtype)[:2]=="fl":
            pass
        else:
            len_of_unique= len(df[column].unique())/df.shape[0]
            if len_of_unique<0.15:
                df[column] = df[column].astype("category")

    return df
                
            
                
    
```
