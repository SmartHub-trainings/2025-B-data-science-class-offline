```python
def determine_independence(row):
    age=row.loc["Age"]
    p_class =row.loc["Pclass"]
    # print(age)
    # print(p_class)
    # print()
    if age <18 and p_class<3:
        return "Dependant"
    return "Self sponsored"
    
```

```python
df.apply(determine_independence,axis=1)
```

```python
df.apply(determine_independence,axis=1).value_counts()
```

```python
df.apply(determine_independence,axis=1).value_counts(normalize=True)
```

```python
df.apply(determine_independence,axis=1).value_counts(normalize=True)*100
```
