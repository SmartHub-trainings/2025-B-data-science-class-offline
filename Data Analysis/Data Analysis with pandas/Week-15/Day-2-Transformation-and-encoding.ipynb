```python
def get_last_name(name):
    return name.split(",")[0]
```

```python
get_last_name("Clark, Miss Mabel")
```

```python
get_last_name("Adams, Miss Rose")
```

```python
df.Name.apply(get_last_name)
```

```python
def get_other_names(name):
    return name.split(",")[-1].strip()
```

```python
get_other_names("Clark, Miss Mabel")
```

```python
df.Name.apply( get_other_names)
```
