```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)

# Create sales data
sales_data = {
    'date': pd.date_range(start='2023-01-01', periods=100, freq='D'),
    'store': np.random.choice(['A', 'B', 'C', 'D'], size=100),
    'product': np.random.choice(['Widget', 'Gadget', 'Tool', 'Device'], size=100),
    'sales': np.random.randint(10, 100, size=100),
    'units': np.random.randint(1, 10, size=100),
    'promotion': np.random.choice([True, False], size=100),
}

sales_df = pd.DataFrame(sales_data)
sales_df['month'] = sales_df['date'].dt.month_name()
sales_df['day'] = sales_df['date'].dt.day_name()

# Create customer data
customers = {
    'customer_id': range(1, 51),
    'store': np.random.choice(['A', 'B', 'C', 'D', 'E'], size=50),
    'age': np.random.randint(18, 80, size=50),
    'gender': np.random.choice(['M', 'F', 'Other'], size=50, p=[0.48, 0.48, 0.04]),
    'membership_level': np.random.choice(['Bronze', 'Silver', 'Gold', 'Platinum'], size=50),
    'signup_date': pd.date_range(start='2022-01-01', periods=50)
}

customer_df = pd.DataFrame(customers)

# Create transaction data that links to customers
transactions = {
    'transaction_id': range(1, 201),
    'customer_id': np.random.choice(range(1, 51), size=200),
    'date': pd.date_range(start='2023-01-01', periods=200, freq='H'),
    'amount': np.random.uniform(10, 500, size=200).round(2),
    'product_category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Home', 'Beauty'], size=200)
}

transaction_df = pd.DataFrame(transactions)

# Create employee data for each store
employees = {
    'employee_id': range(1, 31),
    'store': np.random.choice(['A', 'B', 'C', 'D'], size=30),
    'department': np.random.choice(['Sales', 'Management', 'Support', 'Logistics'], size=30),
    'hire_date': pd.date_range(start='2020-01-01', periods=30, freq='W'),
    'salary': np.random.normal(50000, 15000, size=30).round(2)
}

employee_df = pd.DataFrame(employees)

# Create store information data
store_info = pd.DataFrame({
    'store': ['A', 'B', 'C', 'D', 'E'],
    'location': ['North', 'South', 'East', 'West', 'Central'],
    'size_sqft': [15000, 8000, 5000, 12000, 10000],
    'opening_date': pd.to_datetime(['2018-03-15', '2019-06-25', '2020-11-10', '2017-01-05', '2021-08-30']),
    'manager_id': [5, 12, 18, 24, None]  # Note: Store E has no manager ID
})

# Optional: Save to CSV for reference
sales_df.to_csv('sales_data.csv', index=False)
customer_df.to_csv('customer_data.csv', index=False)
transaction_df.to_csv('transaction_data.csv', index=False)
employee_df.to_csv('employee_data.csv', index=False)
store_info.to_csv('store_info.csv', index=False)

# Display first few rows of each dataset
print("Sales Data:")
print(sales_df.head(3))
print("\nCustomer Data:")
print(customer_df.head(3))
print("\nTransaction Data:")
print(transaction_df.head(3))
print("\nEmployee Data:")
print(employee_df.head(3))
print("\nStore Information:")
print(store_info.head())
```
