Write a program that takes some number from the user , then compute the average of these  numbers.

* The quantity of numbers to be taken should be specified by the user.
* The program should stop taking numbers when the user enter nothing.

```python
# 1.1
number_of_items=int(input("Enter number of items: "))
sum_of_numbers=0
for k in range(number_of_items):
    number = float(input("Enter a number: "))
    sum_of_numbers = sum_of_numbers+ number
avg = sum_of_numbers/number_of_items
print(f"The average is {avg}")
    
```

```python
# 1.1: Method 1
number_of_items=int(input("Enter number of items: "))
sum_of_numbers=0
for k in range(number_of_items):
    number = float(input("Enter a number: "))
    sum_of_numbers += number
avg = sum_of_numbers/number_of_items
print(f"The average is {avg}")
    
```

```python
# 1.1: Method 2
number_of_items=int(input("Enter number of items: "))
numbers=[]
for k in range(number_of_items):
    number = float(input("Enter a number: "))
    numbers.append(number)
print(numbers)
sum_of_numbers=sum(numbers)
avg=sum_of_numbers/number_of_items
print(avg)
    
```

```python
sum_of_numbers=0
count=0
while True:
    number = input("Enter a number: ")
    if number=="":
        break
    sum_of_numbers += float(number)
    count +=1
avg = sum_of_numbers/count
print(f"The average is {avg}")
    
```
