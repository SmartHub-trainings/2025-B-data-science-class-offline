Create a program that prints the Fibonacci sequence up to a specific count provided by the user.

0,1,.....

```python
# number_of_terms= int(input("Enter the number of terms: "))
fib_sequence=[0,1]
for i in range(2,10):
    next_term = fib_sequence[i-1]+fib_sequence[i-2]
    fib_sequence.append(next_term)
print(fib_sequence)
    
```

```python
class BankAccount:
    def __init__(self,bvn,name,phone,email,address,init_dep):
        self.name=name
        self.email=email
        self.address=address
        self.bvn=bvn
        self.contact=phone
        self.initial_deposit=init_dep
        self.balance=init_dep
        
        
    def make_withdrawal(self,amount):
#         self.balance =self.balance-amount
        if amount<=self.balance-1000:
            self.balance -= amount
        else:
            print("Insufficient fund")
```

```python
victory_account =BankAccount("343432432",
                             "victory val",
                             "+2348173433",
                             "vic@gmail.com",
                             "125 Valentine str.",
                             4e3
                            )
```

```python
victory_account.balance
```
